# Day 15 – NumPy Statistical Analysis

## Objective
Analyze a numerical dataset using NumPy statistical functions and identify important patterns in the data.

## Tools Used
- Python
- NumPy
- Jupyter Notebook

```python
import numpy as np
```

```python
marks = np.array([65, 72, 88, 91, 56, 78, 84, 95, 67, 73, 89, 45, 82, 76, 93])

marks
```

```python
mean_marks = np.mean(marks)

print("Mean:", mean_marks)
```

```python
median_marks = np.median(marks)

print("Median:", median_marks)
```

```python
std_marks = np.std(marks)

print("Standard Deviation:", std_marks)
```

```python
minimum_marks = np.min(marks)
maximum_marks = np.max(marks)

print("Minimum:", minimum_marks)
print("Maximum:", maximum_marks)
```

```python
print("Statistical Analysis")
print("--------------------")
print("Mean:", np.mean(marks))
print("Median:", np.median(marks))
print("Standard Deviation:", np.std(marks))
print("Minimum:", np.min(marks))
print("Maximum:", np.max(marks))
```

## Comparison of Mean and Median

The mean represents the average value of the dataset, while the median represents the middle value when the data is arranged in ascending order.

Comparing the mean and median helps us understand whether the dataset is balanced or influenced by unusually high or low values.

```python
if mean_marks > median_marks:
    print("Mean is greater than the median.")
elif mean_marks < median_marks:
    print("Mean is less than the median.")
else:
    print("Mean and median are equal.")
```

## Identifying Unusually High and Low Values

To identify unusual values, we will examine students who scored 90 or above and students who scored below 50.

```python
high_values = marks[marks >= 90]
low_values = marks[marks < 50]

print("High values (90 and above):", high_values)
print("Low values (below 50):", low_values)
```

## Highest and Lowest Performance

```python
highest_mark = np.max(marks)
lowest_mark = np.min(marks)

print("Highest mark:", highest_mark)
print("Lowest mark:", lowest_mark)
```

```python
data_range = np.max(marks) - np.min(marks)

print("Range:", data_range)
```

## Observations

1. The mean and median are relatively close, indicating that the dataset does not have strong skewness.
2. The highest mark is 95, while the lowest mark is 45, giving a range of 50 marks.
3. Three students scored 90 or above, while one student scored below 50.
4. The standard deviation indicates how much the student marks vary from the average.

## Interview Questions

### 1. What is the difference between mean and median?

The mean is calculated by adding all values and dividing by the number of values. The median is the middle value when the data is arranged in ascending or descending order.

### 2. When can median be more useful than mean?

The median is more useful when a dataset contains extreme values or outliers because it is less affected by unusually high or low values.

### 3. What does standard deviation tell us?

Standard deviation measures how much the values in a dataset are spread around the mean. A low standard deviation means the values are closer to the mean, while a high standard deviation means the values are more widely spread.
